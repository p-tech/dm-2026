**Common aggregation functions:**

```python
import pandas as pd

df = pd.DataFrame({
    "department": ["Sales", "Sales", "Tech", "Tech"],
    "employee": ["A", "B", "C", "D"],
    "salary": [40000, 50000, 60000, 70000],
    "sales": [10, 15, 8, 20]
})
```

Aggregate a whole column:

```python
df["salary"].sum()
df["salary"].mean()
df["salary"].median()
df["salary"].min()
df["salary"].max()
df["salary"].count()
df["salary"].std()
```

Will only show the last expression.

```python
df["salary"].agg(["sum", "mean", "median", "min", "max", "count", "std"])
```

Similar to SQL - we can aggregate based on groups.

```python
df.groupby("department")["salary"].agg(["count", "sum", "mean", "min", "max"])
```

Carry out different calculations to each column in the same groupby

```python
df.groupby("department").agg(
    total_salary=("salary", "sum"),
    average_salary=("salary", "mean"),
    total_sales=("sales", "sum"),
    max_sales=("sales", "max")
)
```

Custom aggregations

```python
df.groupby("department").agg(
    salary_range=("salary", lambda x: x.max() - x.min())
)
```

Lamda is a small function without having to declare one explicity in the code, using a def.

```python
def salary_range(x):
    return x.max() - x.min()

df.groupby("department").agg(
    salary_range=("salary", salary_range)
)
```
