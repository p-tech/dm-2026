# Student Task: Extract Weather Data from Open-Meteo

## Goal
Use the Open-Meteo API to download daily weather data for Coventry, UK, and prepare a clean pandas DataFrame. In the next exercise, this DataFrame will be used for calculations, aggregations, and reporting.

**Do not calculate averages, totals, or other summaries in this workbook.**

## Data to retrieve

Location: **Coventry, UK**

- Latitude: `52.4068`
- Longitude: `-1.5197`
- Date range: `2026-01-01` to `2026-01-31`

The API provides daily maximum temperature, minimum temperature, and total precipitation.

## Part 1 — Request the data

Run the cell below. It imports the required libraries, sends the API request, checks that it succeeded, and stores the JSON response in `weather_json`.

```python
import requests
import pandas as pd

url = (
    "https://archive-api.open-meteo.com/v1/archive"
    "?latitude=52.4068&longitude=-1.5197"
    "&start_date=2026-01-01&end_date=2026-01-31"
    "&daily=temperature_2m_max,temperature_2m_min,precipitation_sum"
    "&timezone=Europe%2FLondon"
)

response = requests.get(url)
response.raise_for_status()
weather_json = response.json()
```

```python
# 1. What are the top-level keys in weather_json?
print("Top-level keys:", list(weather_json.keys()))
```

## Part 3 — Create a DataFrame

Create a DataFrame from the API's daily section. It must contain the following columns, with exactly these names:

- `date`
- `max_temperature_c`
- `min_temperature_c`
- `precipitation_mm`

Hint: the data in the daily section is already arranged as columns of equal-length lists. `pd.DataFrame(...)` can convert it into a DataFrame.

```python
# 1. Create a DataFrame from weather_json["daily"].
df = pd.DataFrame(weather_json["daily"])

# 2. Rename its columns to the required names above.
df = df.rename(columns={
    "time": "date",
    "temperature_2m_max": "max_temperature_c",
    "temperature_2m_min": "min_temperature_c",
    "precipitation_sum": "precipitation_mm"
})
```

Task: Calculate the average daily temperature

```python
df["temperature_range_c"] = (
    df["max_temperature_c"] - df["min_temperature_c"]
)

df.head()
```

Questions:

1.   Which date had the largest temperature range?
2.   What was that range, in °C?
3.   Which date had the smallest temperature range?



```python
largest_range_row = df.loc[df["temperature_range_c"].idxmax()]
smallest_range_row = df.loc[df["temperature_range_c"].idxmin()]

print(
    "Largest temperature range: "
    f"{largest_range_row['date'].date()} "
    f"({largest_range_row['temperature_range_c']:.1f} °C)"
)
print(
    "Smallest temperature range: "
    f"{smallest_range_row['date'].date()} "
    f"({smallest_range_row['temperature_range_c']:.1f} °C)"
)
```

Calcualte the monthly summary figures.

```python
mean_max_temp = df["max_temperature_c"].mean()
mean_min_temp = df["min_temperature_c"].mean()
mean_temperature_range = df["temperature_range_c"].mean()
total_precipitation = df["precipitation_mm"].sum()

dry_days = (df["precipitation_mm"] == 0).sum()
wet_days = (df["precipitation_mm"] > 0).sum()

wettest_day = df.loc[df["precipitation_mm"].idxmax()]

print(f"Mean maximum temperature: {mean_max_temp:.1f} °C")
print(f"Mean minimum temperature: {mean_min_temp:.1f} °C")
print(f"Mean daily temperature range: {mean_temperature_range:.1f} °C")
print(f"Total precipitation: {total_precipitation:.1f} mm")
print(f"Dry days: {dry_days}")
print(f"Wet days: {wet_days}")
print(
    f"Wettest day: {wettest_day['date'].date()} "
    f"({wettest_day['precipitation_mm']:.1f} mm)"
)
```

Aggregate the data by week

```python
df["week"] = df["date"].dt.isocalendar().week

weekly_summary = (
    df.groupby("week")
    .agg(
        average_max_temperature_c=("max_temperature_c", "mean"),
        average_min_temperature_c=("min_temperature_c", "mean"),
        total_precipitation_mm=("precipitation_mm", "sum"),
        average_temperature_range_c=("temperature_range_c", "mean"),
    )
    .round(1)
)

weekly_summary
```

Create a monthly report

```python
monthly_report = pd.DataFrame({
    "mean_max_temperature_c": [mean_max_temp],
    "mean_min_temperature_c": [mean_min_temp],
    "mean_temperature_range_c": [mean_temperature_range],
    "total_precipitation_mm": [total_precipitation],
    "dry_days": [dry_days],
    "wet_days": [wet_days],
})

monthly_report.round(1)
```

```python
monthly_report.round(1).to_csv(
    "coventry_january_2026_monthly_report.csv", index=False
)
weekly_summary.to_csv("coventry_january_2026_weekly_summary.csv")
```
